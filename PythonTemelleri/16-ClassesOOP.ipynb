```python
class Dog():
    year = 7

    def __init__(self,age=5):
        self.age = age
        self.dogHumanAge = age * self.year

    def humanAge(self):
        return self.age * self.year # self.year --> Dog.year da yazılabilir!
    
```

```python
dog = Dog(3)
```

```python
dog.humanAge()
```
