# FOR LOOP

```python
myList = [10,20,30,40,50,60,70]
```

```python
for i in myList:
    print(i/5*2)
```

```python
for i in myList:
    if i%6 == 0:
        print(i)
        
```

```python
myString = "gökdeniz tural"
```

```python
for c in myString:
    print(c)
```

```python
my_new_list = [("a","b"),("c","d"),("e","f")]
```

```python
for element in my_new_list:
    print(element)
```

```python
my_new_list[1][1]
```

```python
for (x,y) in my_new_list:
    print(y)
```

```python
myDictionary = {"k1":100, "k2": 200 ,"k3":300}
```

```python
for element in myDictionary:
    print(element)
```

```python
for element in myDictionary.values():
    print(element)
```

```python
for (keys,values) in myDictionary:  # .items() koymalıyız önce sebebi aşağıda
    print(keys,values) 
```

```python
myDictionary.items()   # sözlük içinde tuple'lar oluştu keys ilk değeri values ise 2. değerleri alır
```

```python
for (keys,values) in myDictionary.items():
    print(keys,values)
```

# CONTİNUE - BREAK - PASS

```python
myList = [10,20,30,40,50,60,70]
```

```python
for i in myList:
    print(i)
    if i == 40:
        print("yes")
        break
```

```python
for number in myList:
    if number == 40:
        continue
    print(number)
```

```python
for num in myList:
    pass
```

```python
# pass genellikle bitmiş kodda kullanılmaz. Sonrasında geri dönüp bakmak için geçici kullanırız!
```

# WHİLE LOOP

```python
x = 0
```

```python
while x < 10:
    print(x)
    x += 1
```

```python
myName = "Gökdeniz"
mySurname = "Tural"
```

```python
print(f"Welcome {myName +" "+mySurname}")
```

```python
a = 0
while a < 20:
    print(f"Value of a: {a}")
    a += 1
```
