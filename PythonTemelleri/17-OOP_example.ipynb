# Inheritance

```python
class Musician():
    def __init__(self,name):
         self.name = name
         print("musician class")
    def test1(self):
         print("test1")
    def test2(self):
         print("test2")
```

```python
gokdeniz = Musician("gokdeniz")
```

```python
class MusicianPlus(Musician):
    def __init__(self,name):
        Musician.__init__(self,name)
        print("musician plus")
        
    def test3(self):
        print("test3")

        #override 
    def test1(self):
        print("test1 test1 test1")
```

```python
aleyna = MusicianPlus("Aleyna")
```

```python
aleyna.name = "Aleyna Yildiz"
```

```python
aleyna.name
```

```python
aleyna.test1()
```

```python
type(aleyna)
```

```python
type(gokdeniz)
```

```python
aleyna.test3()
```

```python
aleyna.test1()
```

# Polymorphism

```python
class Musician():
    def play(self):
        print("Müzisyen bir şeyler çaliyor...")

class Guitarist():
    def play(self):
        print("Gitarist solo atiyor!")

class Drummer():
    def play(self):
        print("Davulcu coşturuyor!")
        
```

```python
band = [Guitarist(),Drummer(),Musician()]
```

```python
for member in band:
    member.play()
```

# Encapsulation

```python
class Phone():
    def __init__(self,name,price):
        self.name = name
        self.__price = price # __  --> private tanımlaması

    def info(self):
        print(f"{self.name} price is {self.__price}")

    def changePrice(self,price):
        self.__price = price     # ---> Erişimci metot encapsulation'ın babası!
```

```python
iPhone = Phone("iPhone 14",500)
```

```python
iPhone.info()
```

```python
iPhone.__price = 400
```

```python
iPhone.info()
```

```python
iPhone.changePrice(150)
```

```python
iPhone.info()
```

# Abstraction

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):
    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
#myCar = Car()  # abstract sınıftan nesne oluşturamıyorduk!
```

```python
class Tesla(Car):
    
    def maxSpeed(self):
        print("200km")
```

```python
tesla = Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes(Car):

    def maxSpeed(self):
        print("300km")
```

```python
mercedes = Mercedes()
```

```python
mercedes.maxSpeed()
```
