```python
def divideNumber(number):
    return number / 2
```

```python
myList = [3,5,7,10,20,30]
```

```python
myResultList = []
```

```python
for num in myList:
    myResultList.append(divideNumber(num))
```

```python
myResultList
```

# map

```python
list(map(divideNumber,myList))
```

```python
def controlString(string):
    return "Gokdeniz" in string
```

```python
controlString("Ahmet")
```

```python
myStringList = ["Gokdeniz","Aleyna","Tolga","Ali","Furkan","Gokdeniz Tural"]
```

```python
list(map(controlString,myStringList))
```

# filter

```python
list(filter(controlString,myStringList))
```

# SCOPE

```python
#Global
myName = "Gokdeniz" 
def nameFunction():
    #Enclosing
    myName = "Gokdeniz2"
    print(myName)

    def nameFunction2():
        #Local
        myName = "Gokdeniz3"
        print(myName)
    nameFunction2()
```

```python
nameFunction()
```

```python
myName
```

```python
y = 10

def degistir(y):
    print(y)
    y = 5
    print(y)
```

```python
y
```

```python
degistir()
```

```python
y = 10

def degistir():
    global y
    y = 5
    print(y)
```

```python
degistir()
```
