```python
class Fruit():
    def __init__(self,name,calories):
        self.name = name
        self.calories = calories

    def __str__(self):
        return f"{self.name}: {self.calories} calories"
    
    def __len__(self):
        return self.calories
```

```python
myFruit = Fruit("banana",150)
```

```python
myFruit.calories
```

```python
myFruit.name
```

```python
print(myFruit)  # Objeyi print ettirdiğimizde __str__ metodu ile bilgileri yazdırabildik. Direkt obje print ettik!
```

```python
len(myFruit)
```

```python
class Student():
    def __init__(self,name):
        self.name = name
    def __getitem__(self,key): # getitem methodunu override edip özelleştirdik ve böyle bir uygulama yaptık. Daha fazlası için dökümanlara bak!
        if key == "gokdeniz":
            return self.name
        else:
            return "Not found!"
```

```python
st = Student("gokdeniz")
```

```python
st["gokdeniz"]
```
