```python
#4) Aşağıdaki kodun çıktısı ne olacaktır?
```

```python
def myLoop(*args):  # *args yazdığımızda istediğimiz kadar değer verebiliyorduk
    for element in args:
        print(element / 2)
```

```python
myLoop(3,2,1,5,3,4,6,10)
```

```python
#5) Aşağıdaki dizide belirtilen rakamları, myFunction fonksiyonuna tabi tutup, yeni bir dizi oluşturunuz
```

```python
def myFunc(num):
    return num ** 3
```

```python
myList = [2,3,4,5,6]
```

```python
newList = []
```

```python

for number in myList:
    x = myFunc(number)
    newList.append(x)
```

```python
newList
```

```python
list(map(myFunc,myList))  # 2. yol ve basit yolu!
```

```python
#6) Aşağıdaki string dizisinde, içinde sadece XYZ geçen barkodları gösterecek yeni bir liste oluşturunuz
```

```python
barkodDizisi = ["ABC231","SA3123XYZ","XYZA123Q","QRE1231KJ","X112QGL"]
```

```python
newBarkodList = []

for bar in barkodDizisi:
    if "XYZ" in bar:
        newBarkodList.append(bar)

newBarkodList
```

```python
list(filter(lambda string : "XYZ" in string,barkodDizisi)) # İKİNCİ YOL!
```

```python
#8) Aşağıda yazdırılan sınıfı incelediğinizde kedim.yasiCarp() kodunun çıktısı ne olacaktır?
```

```python
class Kedi():
        
    def __init__(self,isim,yas=5):
        self.isim = isim
        self.yas = yas
        
    def yasiCarp(self):
        return self.yas * 3
```

```python
kedim = Kedi("Tonton")
```

```python
kedim.yasiCarp()
```

```python
#9) Aşağıdaki kodun çıktısı ne olacaktır?
```

```python
class Ogrenci():
    
    def __init__(self,isim,sinavNotu):
        self.isim = isim
        self.__sinavNotu = sinavNotu
    
    def notuGoster(self):
        print(f"{self.isim} sınav notu: {self.__sinavNotu}")
```

```python
ogrenci = Ogrenci("Mehmet",85)
```

```python
ogrenci.__sinavNotu = 75
```

```python
ogrenci.notuGoster()
```
