```python
myList = [10,20,30,40,50]
```

```python
for ix in range(len(myList)):
    print(ix)
```

```python
#enumerate
```

```python
for element in enumerate(myList):
    print(element)
```

```python
for (ix,element) in enumerate(myList):
    print(ix)
    print(element)
```
