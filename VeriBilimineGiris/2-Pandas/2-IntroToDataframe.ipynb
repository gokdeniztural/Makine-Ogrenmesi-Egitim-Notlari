```python
# MULTİ İNDEX
```

```python
first_index = ["Simpson","Simpson","Simpson","South Park","South Park","South Park",]
second_index = ["Homer","Bart","Marge","Cartman","Kenny","Kyle"]
```

```python
zipped_index = list(zip(first_index,second_index))
```

```python
zipped_index
```

```python
zipped_index = pd.MultiIndex.from_tuples(zipped_index)
```

```python
zipped_index
```

```python
# Şimdi bunu kullanarak data frame oluşturabiliriz...
```

```python
sample_values = np.ones((6,2))
```

```python
big_df = pd.DataFrame(sample_values , index = zipped_index , columns = ["Age","Salary"])

# BURASI ÇOK ÖNEMLİ !!!!!!!
```

```python
big_df
```
