```python
# İki ayrı ama benzer datalar içeren data frameleri birleştireceğiz bunu da 'concat' dediğimiz fonksiyon aracılığı ile yapıyoruz.
```

# concat

```python
df_concat = pd.concat([df1,df2], ignore_index = True)

# ignore_index --> indexleri göz ardı et. İki dataframe indeksleri de 0-9 arası ama biz bunları birleştireceğimiz için göz ardı etmeliyiz o yüzden
# ignore_index değerine True vermeliyiz...
```

```python
df_concat
```

# merge

```python
df_merge1 = pd.read_csv('7-merge_data1.csv')
```

```python
df_merge1
```

```python
df_merge2 = pd.read_csv('7-merge_data2.csv')
```

```python
df_merge2
```

```python
# merge - outer join
```

```python
df_merged_outer = pd.merge(df_merge1,df_merge2, on = "Employee_ID", how = "outer")
```

```python
df_merged_outer
```

```python
# outer her şeyi kapsıyor. İki tabloyu birleştiriyor match edebildiklerini (ID ' ye göre ) alıp yan yana yazdırıyor ve bütün dataları saklıyor...
```

```python
# merge - left join
```

```python
df_merged_left = pd.merge(df_merge1,df_merge2, on = "Employee_ID", how = "left")
```

```python
df_merged_left
```

```python
# 2. tabloyu göz ardı ediyor. 1. tablonun salary ve experience değerlerini ikince tabloda ID ile uyumlu ise matchleyip taşıyor!!! 
# Diğer 11,12,13 numaralı ID değerleri 1. tabloda matchlenmediği için göz ardı ediliyor...
```

```python
df_merged_right = pd.merge(df_merge1,df_merge2, on = "Employee_ID", how = "right")
```

```python
df_merged_right
```

```python
# right da left ile aynı mantık .
```

```python
# merge - inner join
```

```python
df_merged = pd.merge(df_merge1,df_merge2, on = "Employee_ID", how = "inner")
```

```python
df_merged
```

```python
# Her iki tabloda da olan kesişim kümesini alır diğerlerini çıkarır ---> 'inner'
```
