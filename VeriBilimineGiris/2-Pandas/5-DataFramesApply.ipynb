```python
def salary_category(salary):
    if salary < 50000:
        return "Low"
    elif 50000 <= salary < 80000:
        return "Medium"
    else:
        return "High"
```

```python
# Bu fonksiyonu her ögeye uygulayıp yeni bir kolon oluşturup bu kolonlara gerekli return değerlerini döndürmek için "apply" fonksiyonunu kullanırız!!!!
```

```python
df["Salary Category"] = df["Salary"].apply(salary_category) 

# Salary Category adında bir kolon oluşturduk ve bu kolona Salary değerlerini fonksiyonla apply ederek kaydettik!
```

```python
# SORU

# if experience > 10 ----> performance_score + 1
```

```python
def new_score(experience):
    if experience > 10:
       return  1
    else:
        return 0
```

```python
df["Adjusted"] = df["Experience"].apply(new_score)
```

```python
df
```

```python
df["New Score"] = df["Adjusted"] + df["Performance_Score"]
```

```python
df
```

```python
# 2. ÇÖZÜM İLERİDE GEREKLİ OLABİLİR !!!!!!!!!!
```

```python
def adjusted_score(row):
    if row["Experience"] > 10:
        return row["Performance_Score"] + 1
    else:
        return row["Performance_Score"]
```

```python
df["Adjusted_score"] = df.apply(adjusted_score, axis = 1)
```

```python
df
```

```python
# Name değerlerindedki örn: Emp_1 olan ifadedeki "_" yi kaldırıp Emp 1 olarak değiştirmek isteyelim. Bunun için fonksiyon yazma lambda fonksiyon kullan.
```

```python
df["Formatted_Name"] = df["Name"].apply(lambda x: x.replace("_"," "))
```
