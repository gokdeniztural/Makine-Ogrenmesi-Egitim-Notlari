# working with missing data

```python
weather_na_df = pd.read_excel('6-weatherna.xlsx')
```

```python
weather_na_df
```

```python
weather_na_df.isna()
```

```python
weather_na_df.describe()
```

```python
weather_na_df['Paris'].count()
```

```python
weather_na_df
```

```python
weather_na_df.dropna() # --> axis = 0

# Eksik dataya sahip rowları dahil etmez!
```

```python
weather_na_df
```

```python
weather_na_df.drop('Paris', axis = 1)
```

```python
weather_na_df.fillna(19)

# Missing Data NaN değerlerini istediğimiz bir değerle doldurur!
```

```python
weather_na_df.fillna(weather_na_df.mean())  # NaN değerlerine her şehrin ortalama değerini doldurur!
```

```python
weather_na_df.mean() # ortalamalar NaN değerlerinin yerlerine girildi!
```

```python
df = pd.read_csv("6-employee.csv")
```

```python
df
```

```python
df.describe() # sayısal değerleri , açıklayabildiği değerleri gösterebiliyor.
```

```python
# df.mean() --> çalışmaz. Numerik olmayan datalar mevcut. ama şu şekilde çalıştırabiliriz ...

df['Salary'].mean()
```

```python
print(df[['Salary','Experience']].mean())
```

```python
df.head()
```

```python
df.groupby('Department')  # ---> Departmanlara göre grupluyor ama bunu başka bir değişkene kaydetmek daha iyi olacaktır.
```

```python
df_grouped = df.groupby('Department')
```

```python
df_grouped.count()
```

```python
df_grouped.describe()
```

```python
df_grouped['Salary'].mean()
```

```python
df_city = df.groupby('City')
```

```python
df_city['Salary'].mean()
```
