```python
df_app_category = df_clean.groupby(["Category", "App"])["Installs"].sum().reset_index()
```

```python
# Burada Installs'a göre en çok indirilenden itibaren dizdik. Ama biz apps listesindeki top 5 category'e göre istiyoruz buna uymayanlar olmayacak!
```

```python
apps = ["GAME", "COMMUNICATION", "TOOLS", "PRODUCTIVITY", "SOCIAL"] # top 5 kategorimiz bunlar.

plt.figure(figsize=(30,20))
for i,app in enumerate(apps):
    df2 = df_app_category[df_app_category['Category'] == app] # "Category sütununda 'GAME' yazanları bul ve o satırların tamamını getir"!
    df2 = df2.head(5)
    #print(df2)

    plt.subplot(3,2,i+1)
    sns.barplot(data=df2, x="Installs", y="App", color = "tomato")
    plt.title(app, size=20)
   
plt.tight_layout()
plt.show()
```
