```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('athlete_events.csv')
```

```python
data.describe()
```

```python
data.head()
```

```python
data.isnull()
```

```python
data.info()
```

```python
plt.scatter("Height","Weight", data = data, color = "red")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Athlete's Height vs Weight")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.scatterplot(x="Height", y="Weight",data=data) # x ve y parametrelerine datamızdaki Height ve Weight değerlerini verdik!
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Height vs Weight")
plt.show()
```

```python
# Matplotlib ve seaborn'ü birlikte kullanarak yaptık. Seaborn görselleştirmemizi iyileştirmek için kullanmamız gereken bir kütüphane.
```

```python
sns.set_style("darkgrid") # dark gridde arka daha koyu. Grid ölçek çizgileri ekliyor. (mazgallı yapı).
sns.scatterplot(x="Height", y="Weight",data=data) # x ve y parametrelerine datamızdaki Height ve Weight değerlerini verdik!
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Height vs Weight")
plt.show()
```

```python
# Seaborn güzelliği ---> Mesela cinsiyete göre boy ve kilo nasıl dağılıyor görmek istiyoruz.

sns.set_style("whitegrid")
sns.scatterplot(x="Height", y="Weight",hue = "Sex", style="Medal", data=data) # style "medal" --> madalyaları da simgeliyor.
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Height vs Weight")
plt.show()

# Kadınlar Turuncu
# Erkekler Mavi ---> istediğimiz sonucu aldık
```

```python
data["Medal"].unique() # pandasla Medal kolonunda geçen unique değerleri görebiliyoruz.
```

```python
data["Sex"].unique()
```

```python
sns.set_style("whitegrid")
sns.scatterplot(x="Height", y="Weight",hue = "Sex", style="Medal", size="Age", data=data) # style "medal" --> madalyaları da simgeliyor.
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Height vs Weight")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.lineplot(x="Height",y="Weight",hue="Sex",data=data) # lineer grafik!
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Height vs Weight")
plt.show()
```

```python
sns.set_style("white")
sns.displot(x="Height", hue = "Sex", data=data)
plt.ylabel("Frequency")
plt.title("Atheletes Height vs Weight")
plt.show()
```

```python
sns.set_style("white")
sns.displot(x="Height", hue = "Sex", data=data, kind = "kde") # kind = "kde" --> çizgi grafiği şeklinde sunar.
plt.ylabel("Frequency")
plt.title("Atheletes Height vs Weight")
plt.show()
```

```python
sns.barplot(x="Medal", y="Height", hue = "Sex", data = data, palette={"M":"lightgreen","F":"pink"})
plt.title("Medals by Height")
plt.show()

# palette ile renkleri ayarlayabiliyoruz!
```
