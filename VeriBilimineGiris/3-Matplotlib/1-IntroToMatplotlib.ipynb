```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
age_list = [15,20,20,20,30,40,50,60,70,45]
weight_list = [35,55,55,67,71,75,78,80,85,105]
```

```python
plt.plot(age_list,weight_list, "r") # r --> red {b,g,y}
#plt.savefig('test.png') ---> kaydeder !

plt.xlabel("Age")
plt.ylabel("Weight")
plt.title("Age vs Weight")

plt.show() # inline olarak gösterir!
```

```python
# numpy ile çalıştırmak
```

```python
np_age_list = np.array(age_list)
np_weight_list = np.array(weight_list)
```

```python
plt.plot(np_age_list,np_weight_list, "g")
plt.title("Age vs Weight")
plt.xlabel("Age")
plt.ylabel("Weight")

plt.show()
```

```python
numpy_arr1 = np.linspace(0,10,20) # 0 dan başla 10 a kadar eşit aralıklarla 20 tane sayı oluştur!
```

```python
numpy_arr1
```

```python
numpy_arr2 = numpy_arr1 ** 3
```

```python
numpy_arr2
```

```python
plt.plot(numpy_arr1,numpy_arr2, "r*-")
plt.show()
```

```python
# 1. dizimiz x ekseni 2. dizimiz y ekseni görüldüğü üzere.
```

```python
plt.subplot(1,2,1) # 1 row 2 kolon 1. grafiği çiziyoruz.
plt.plot(numpy_arr1,numpy_arr2,"r*-")
plt.subplot(1,2,2) # 1 row 2 kolon 2. grafiği çiziyoruz.
plt.plot(numpy_arr2,numpy_arr1, "g--")
plt.show()

# plt.subplot ---> tek pencerede yan yana iki grafik oluşturur!
```

```python
my_figure =  plt.figure()
figure_axes = my_figure.add_axes([0.1,0.1,0.6,0.6])
figure_axes.plot(numpy_arr1,numpy_arr2, "g")
figure_axes.set_xlabel("X axis")
figure_axes.set_ylabel("Y axis")
figure_axes.set_title("Graph Title")
plt.show()
```

```python
new_fig = plt.figure(dpi = 100)
new_axes = new_fig.add_axes([0.1,0.1,0.9,0.9])

new_axes.plot(numpy_arr1 , numpy_arr1 ** 2 , label = "numpy array ** 2")
new_axes.plot(numpy_arr1 , numpy_arr1 ** 3 , label = "numpy array ** 3")
new_axes.legend() # label gösterio konumunu ayarlayabiliyoruz (loc = 1,2,3...)
plt.show()
```
