```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0,10,20)
```

```python
data1
```

```python
data2 = data1 ** 2
```

```python
data2
```

```python
plt.subplots() # --> figür (figure) + eksen (axes) oluşturur.
```

```python
my_fig , my_axes = plt.subplots() # burada figüre ve eksene atamalar yapıyoruz daha rahat özel kullanmak için
```

```python
type(my_fig)
```

```python
type(my_axes)
```

```python
my_fig , my_axes = plt.subplots()
my_axes.plot(data1,data2, color = "#000000", alpha = 0.5)
my_axes.plot(data2,data1, color = "#13AE80")
plt.show()

# alpha saydamlık sağlıyor.
```

```python
new_fig , new_axes = plt.subplots()

new_axes.plot(data1, data1 + 2, color="blue", linewidth = 1.0)
new_axes.plot(data1, data1 + 3, color="red", linewidth = 1.0 , linestyle = ":")
new_axes.plot(data1, data1 + 4, color="purple", linewidth = 1.0, linestyle = "-.")
new_axes.plot(data1, data1 + 5, color="black", linewidth = 1.0, linestyle = "-." , marker="o", markersize = 8 , markerfacecolor = "red")
new_axes.plot(data1, data1 + 6, color="purple", linewidth = 1.0, linestyle = "-." , marker = "x", markersize = 10)



plt.show()
```
