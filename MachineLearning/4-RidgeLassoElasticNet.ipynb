```python
# redundancy -> iki farklı kolonun birbiriyle yüksek kolerasyonda olması model için gereksiz karmaşıklığa sebebiyet veriyor !!!!
# multicollinearity -> Çoklu doğrusal bağıntı. , overfitting... 

# Kolerasyonu yüksek olan değişkenleri çıkarmak bizim için mantıklı olacaktır!!! %85, %95 -> Bu civarda kolerasyonları elemek mantıklı!
```

```python
X_train.corr().iloc[0,2]
```

```python
def correlation_for_dropping(df,threshold):
    columns_to_drop = set()
    corr = df.corr()
    for i in range(len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i,j]) > threshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop


# Hangi kolonları atacağımızı görmemiz için oluşturduğumuz bir fonksiyon!! mantığını iyi anla.
```

```python
columns_dropping = correlation_for_dropping(X_train,0.85)
```

```python
columns_dropping
```

```python
X_train.drop(columns_dropping, axis=1, inplace=True)
X_test.drop(columns_dropping, axis=1, inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
plt.subplots(figsize =(15,5))
plt.subplot(1,2,1)
sns.boxplot(data = X_train)
plt.title("X_train")
plt.subplot(1,2,2)
sns.boxplot(data = X_test_scaled)
plt.title("X_test_scaled")
plt.show()
```

```python
# scale olayının ne işe yaradığını gözlemleyerek anlamak için yapıldı ....
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
linear = LinearRegression()
linear.fit(X_train_scaled, y_train)

y_pred = linear.predict(X_test_scaled)

mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test,y_pred)

print("Mean Absolute Error: ", mae)
print("Mean Squared: ", mse)
print("R2 Score: ",r2)
plt.scatter(y_test,y_pred, color = "r")
plt.show()
```

```python
from sklearn.linear_model import Lasso

lasso = Lasso()
lasso.fit(X_train_scaled, y_train)

y_pred = lasso.predict(X_test_scaled)

mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test,y_pred)

print("Mean Absolute Error: ", mae)
print("Mean Squared: ", mse)
print("R2 Score: ",r2)
plt.scatter(y_test,y_pred, color = "r")
plt.show()
```

```python
from sklearn.linear_model import Ridge

ridge = Ridge()
ridge.fit(X_train_scaled, y_train)

y_pred = ridge.predict(X_test_scaled)

mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test,y_pred)

print("Mean Absolute Error: ", mae)
print("Mean Squared: ", mse)
print("R2 Score: ",r2)
plt.scatter(y_test,y_pred, color = "r")
plt.show()
```

```python
from sklearn.linear_model import ElasticNet

elastic = ElasticNet()
elastic.fit(X_train_scaled, y_train)

y_pred = elastic.predict(X_test_scaled)

mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test,y_pred)

print("Mean Absolute Error: ", mae)
print("Mean Squared: ", mse)
print("R2 Score: ",r2)
plt.scatter(y_test,y_pred, color = "r")
plt.show()
```
