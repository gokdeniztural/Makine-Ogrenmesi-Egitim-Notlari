```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('6-bank_customers.csv')
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
X = df.drop("subscribed", axis=1)
y = df["subscribed"]
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
score = accuracy_score(y_pred, y_test) # Tüm doğru tahminlerin toplam tahmin sayısına oranı:
print("score:",score)                  # Accuracy = (TP + TN) / (TP + TN + FP + FN)

print(classification_report(y_pred,y_test))
print("confusion_matrix: \n", confusion_matrix(y_pred,y_test))


# Precision (Kesinlik)
# Pozitif olarak tahmin edilenlerden kaçının gerçekten pozitif olduğu: ----> Precision = TP / (TP + FP)

# Recall (Duyarlılık/Sensitivity)
# Gerçek pozitiflerden kaçının doğru tahmin edildiği: ----> Recall = TP / (TP + FN)

# F1-Score
# Precision ve Recall'ın harmonik ortalaması: ----> F1-Score = 2 × (Precision × Recall) / (Precision + Recall)
# Avantajı: Hem precision hem recall'ı dengeler, özellikle dengesiz veri setlerinde faydalı.
```

```python
# confusion matrixde sol üst ve sağ alttaki değerler daha fazla olduğunu görebiliyoruz bu sebeple accuracy bayağı yüksek çıktı. 11 ve 13 olan ise
# yanılsamalardır.

#Hangi Metriği Ne Zaman Kullanmalı?

# Accuracy: Dengeli veri setlerinde genel performans için
# Precision: Yanlış pozitiflerin maliyeti yüksekse (örn: spam filtresi)
# Recall: Yanlış negatiflerin maliyeti yüksekse (örn: hastalık tespiti)
# F1-Score: Dengesiz veri setlerinde veya precision-recall dengesinde
```

```python
# hyperparameter tuning
```

```python
model = LogisticRegression()
```

```python
penalty = ["l1", "l2", "elasticnet"] # --> cezalar
c_values = [100, 10 , 1, 0.1, 0.01] # --> c değerleri , penalty'lerin lambda oranları yani ne kadar güçlü bir ceza vericez onu gösteriyor. default = 1
solver = ["newton-cg", "lbfgs", "liblinear", "sag", "saga", "newton-cholesky"]
```

```python
parameters = dict(penalty=penalty, C=c_values, solver=solver) # sözlüğe çevirdik yukarıdakileri
```

```python
parameters
```

```python
# grid search cv
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
cv = StratifiedKFold() # ---> cross validation
```

```python
grid = GridSearchCV(estimator = model, param_grid = parameters, cv = cv, scoring = "accuracy", n_jobs = -1 ) # ---> estimator : tahminci
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred, y_test)
print("score: ",score)
print(classification_report(y_pred, y_test))
print("confusion matrix: \n", confusion_matrix(y_pred, y_test))
```

```python
# Grid Search ne yapıyor?

# Sistematik deneme: Her kombinasyonu deniyor
# Cross-validation: Her kombinasyonu 5 farklı train/validation setinde test ediyor
# En iyi skor: En yüksek ortalama CV skorunu buluyor

# -*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*-*
```
